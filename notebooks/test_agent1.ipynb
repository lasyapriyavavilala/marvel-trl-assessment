```python
import sys
sys.path.append('..')

from src.agents.agent_1_decomposer import TechnologyDecomposer
from dotenv import load_dotenv
import os

load_dotenv()
api_key = os.getenv("ANTHROPIC_API_KEY")
```

```python
decomposer = TechnologyDecomposer(api_key=api_key)
```

```python
ontology = decomposer.decompose_reactor("MARVEL")
```

```python
print(f"Reactor: {ontology.reactor}")
print(f"Nodes: {len(ontology.nodes)}")
print(f"Edges: {len(ontology.edges)}")
```

```python
ontology.save("../data/ontologies/marvel_ontology.json")
print("✅ Saved to data/ontologies/marvel_ontology.json")
```

```python
G = ontology.to_networkx()
import matplotlib.pyplot as plt
import networkx as nx

plt.figure(figsize=(12, 8))
pos = nx.spring_layout(G, k=0.5, iterations=50)
nx.draw(G, pos, with_labels=True, node_color='lightblue', 
        node_size=500, font_size=8, arrows=True)
plt.title(f"{ontology.reactor} Ontology Graph")
plt.savefig("../data/marvel_graph.png", dpi=150, bbox_inches='tight')
plt.show()
```
